```python
# lets do a function!
# Changing celsius to fahrenheit.

InitCtemp=0
Ftemp=(9/5)*float(InitCtemp)+32
print(Ftemp)
```

```python
# All commands are basically funcitons, so lets make our own command!
# To do that, let's make a funciton.
   
def temp_cov(temp, scale): #temp and scale are inputs
    if scale.upper()=="C" or scale.upper()=='CELSIUS':
        print(str(temp)+" C = "+str((9/5)*float(temp)+32)+" F")
    elif scale.upper()=='F' or scale.upper()=='FAHRENHEIT':
        print(str(temp)+' F = '+str((5/9)*(float(temp)-32))+' C')
    else:
        print("You made a mistake")
```

```python
temp_cov(100, 'c')
```
